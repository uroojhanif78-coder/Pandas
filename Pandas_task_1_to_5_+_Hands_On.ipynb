```python
# task 1: "create students data frame"

import pandas as pd

data = {
    "Name" : ["ali","sara","ahmed","ayesha","zain"],
    "Marks" :[70,85,60,95,45]
}
df = pd.DataFrame(data)
print(df)
Marks=df["Marks"]
print(Marks.mean())
print(Marks.max())
print(Marks.min())


name =df["Name"]
print(name.count())
```

```python
# task 2: "students whose marks are greater & less than 50"

print(df[df['Marks'] >= 50])
print(df[df['Marks'] <= 50])
```

```python
# task 3: "Add a new column"

df["cities"]=["karachi","lahore","islamabad","faislabad","rawalpindi"]
print("only Karachi Students:")
print(df[df['cities'] == "karachi"])
```

```python
# task 4: "Find the average marks of each course using groupby()"

data = {

    "Name": ["Ali","sara","ahmed","ayesha","zain"],
    "Course":["AI","Web","AI","Web","AI"],
    "Marks":[70,60,80,50,90]

}
df = pd.DataFrame(data)
result=df.groupby("Course")["Marks"].mean()
print(result)
```
