```python
# Selecting column
print(df["population"])
```

```python
# Selecting multiple columns
print(df[["longitude","latitude"]])
```

```python
# selecting rows loc & iloc
# loc
print(df.loc[0])
```

```python
# specific value

print(df.loc[1,"households"])
```

```python
#iloc
print(df.iloc[0, 3])
```

```python
#Adding a new column

df["population_per_household"]= df["population"] / df["households"]
print(df[["population","households","population_per_household"]])
```

```python
# updating values
df.loc[1," housing_median_age"] = 10
print(df.loc[1])
```

```python
#update an entire column
df["total_rooms"] = df["total_rooms"] + 10
print(df["total_rooms"].head())
```

```python
#missing value

print(df.isnull())
```

```python
# fill missing values
df["housing_median_age"] = df["housing_median_age"].fillna(0)
print(df["housing_median_age"].isnull())
```

```python
print(df.duplicated())
```

```python
#Sorting values
# largest to smallest
print(df.sort_values("latitude", ascending=False))
```

```python
# smallest to largest
print(df.sort_values("latitude", ascending=True))
```

```python
#groupby
result = df.groupby("total_bedrooms")["total_rooms"].mean()
print(result)
```

```python
# concat()

df1 = pd.DataFrame({
    "Name": ["faiza","ali","samina","fahad"],
    "Marks" : [90,70,80,50]
    })
df2 = pd.DataFrame({
    "Name": ["amhed","faisal","shiza","sania"],
    "Marks" :[70,95,68,89]
})
final_df = pd.concat([df1, df2], ignore_index=True)
print(final_df)
```
